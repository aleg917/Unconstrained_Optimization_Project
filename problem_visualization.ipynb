# NO4LSP Assignment 2.1 — Test-Problem Visualization

Landscapes of the two test problems for $n = 2$:

- **Problem 16** — Banded Trigonometric (non-convex, several local minima)
- **Problem 28** — Variably Dimensioned (convex, unique global minimum)

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

from src.functions.problem16 import f16
from src.functions.problem28 import f28, x_bar_28
```

## Problem 16 — Banded Trigonometric

$$F(x) = \sum_{i=1}^{n} i \cdot \left[(1 - \cos x_i) + \sin x_{i-1} - \sin x_{i+1}\right]$$

with $x_0 = x_{n+1} = 0$ (fixed boundaries) and starting point $\bar{x}_i = 1$.
The function is non-convex and has many local minima.

```python
# Problem 16 landscape for n=2 (wide range to expose several local minima).
x_lo, x_hi = -2 * np.pi, 2 * np.pi
grid = np.linspace(x_lo, x_hi, 400)
X1, X2 = np.meshgrid(grid, grid)
Z = np.vectorize(lambda a, b: f16(np.array([a, b])))(X1, X2)

# Analytic local minima (the problem is separable for n=2):
#   dF/dx1 = sin x1 + 2 cos x1 = 0  ->  tan x1 = -2,  with cos x1 > 0
#   dF/dx2 = 2 sin x2 - cos x2 = 0  ->  tan x2 = 1/2, with 2 cos x2 + sin x2 > 0
x1_base = np.arctan2(-2.0, 1.0)   # branch with cos > 0
x2_base = np.arctan2(1.0, 2.0)
ks = np.arange(-3, 4)
x1_mins = (x1_base + 2 * np.pi * ks)
x2_mins = (x2_base + 2 * np.pi * ks)
x1_mins = x1_mins[(x1_mins >= x_lo) & (x1_mins <= x_hi)]
x2_mins = x2_mins[(x2_mins >= x_lo) & (x2_mins <= x_hi)]
X1m, X2m = np.meshgrid(x1_mins, x2_mins)

fig, ax = plt.subplots(figsize=(8, 7))
cf = ax.contourf(X1, X2, Z, levels=50, cmap='viridis')
ax.contour(X1, X2, Z, levels=20, colors='white', linewidths=0.4, alpha=0.4)
ax.plot(1, 1, 'ro', markersize=8, label=r'$\bar{x} = (1, 1)$')
ax.scatter(X1m.ravel(), X2m.ravel(), c='cyan', s=80, marker='*',
           edgecolor='black', linewidths=0.8, label='local minima', zorder=5)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_title('Problem 16 — Banded Trigonometric, n=2')
plt.colorbar(cf, ax=ax, label='F(x)')
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

## Problem 28 — Variably Dimensioned

$$F(x) = \tfrac{1}{2}\left[\sum_{k=1}^{n} (x_k - 1)^2 + S^2 + S^4\right],
\qquad S = \sum_{i=1}^{n} i\,(x_i - 1)$$

Convex with a unique global minimum at $x^* = (1, \ldots, 1)$, $F^* = 0$.
Starting point $\bar{x}_l = 1 - l/n$.

```python
# Problem 28 landscape for n=2, window centered on the global minimum x* = (1, 1).
x_lo, x_hi = -1.0, 3.0
grid = np.linspace(x_lo, x_hi, 400)
X1, X2 = np.meshgrid(grid, grid)
Z = np.vectorize(lambda a, b: f28(np.array([a, b])))(X1, X2)

x_star = np.array([1.0, 1.0])   # global minimum (F = 0)
x0_bar = x_bar_28(2)            # = (0.5, 0.0)

fig, ax = plt.subplots(figsize=(8, 7))
# Log scale: F grows like ~ S^4 along the direction j = (1, 2), so on a linear
# scale the level sets around x* would be squashed against the minimum.
cf = ax.contourf(X1, X2, np.maximum(Z, 1e-6), levels=50,
                 norm=LogNorm(vmin=1e-3, vmax=Z.max()), cmap='plasma')
ax.contour(X1, X2, Z, levels=[0.1, 1, 5, 20, 100, 500, 2000],
           colors='white', linewidths=0.5, alpha=0.5)
ax.plot(x0_bar[0], x0_bar[1], 'ro', markersize=8, label=r'$\bar{x} = (0.5,\, 0)$')
ax.scatter([x_star[0]], [x_star[1]], c='cyan', s=120, marker='*',
           edgecolor='black', linewidths=0.8, label='global minimum (F=0)', zorder=5)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_title('Problem 28 — Variably Dimensioned, n=2')
plt.colorbar(cf, ax=ax, label='F(x)  (log scale)')
ax.legend(loc='upper left')
plt.tight_layout()
plt.show()
```
